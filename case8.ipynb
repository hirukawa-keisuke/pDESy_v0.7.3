````python
mother_ship = project.create_product('Mother Ship')

maneuver_sys = mother_ship.create_component('Maneuvering System')
eng_sys = mother_ship.create_component('Engine System')
elec_sys = mother_ship.create_component('Electric System')

for i in range(number_of_days):

    exec(f"maneuver_day{i+1} = project.create_workflow('Maneuver Day{i+1}')")
    shifts = ['04_1', '48_1', '80_1', '04_2', '48_2', '80_2']
    for j, t in enumerate(shifts):
        exec(f"maneuver_{t}_day{i+1}_A = maneuver_day{i+1}.create_task('Maneuver {t} Day{i+1} A', need_facility=True, default_work_amount=240)")
        exec(f"maneuver_{t}_day{i+1}_B = maneuver_day{i+1}.create_task('Maneuver {t} Day{i+1} B', need_facility=True, default_work_amount=240)")
        exec(f"maneuver_sys.update_targeted_task_set({{maneuver_{t}_day{i+1}_A, maneuver_{t}_day{i+1}_B}})")
        exec(f"bridge.update_targeted_task_set({{maneuver_{t}_day{i+1}_A, maneuver_{t}_day{i+1}_B}})")
        exec(f"tasks_maneuver_t1.append(maneuver_{t}_day{i+1}_A,)")
        exec(f"tasks_maneuver_t2.append(maneuver_{t}_day{i+1}_B,)")
        exec(f"maneuver_console_A.workamount_skill_mean_map.update({{maneuver_{t}_day{i+1}_A.name : 1.0,}})")
        exec(f"maneuver_console_B.workamount_skill_mean_map.update({{maneuver_{t}_day{i+1}_B.name : 1.0,}})")
        exec(f"maneuver_{t}_day{i+1}_B.update_input_task_set({{maneuver_{t}_day{i+1}_A}}, BaseTaskDependency.SS)")
        # exec(f"maneuver_{t}_day{i+1}_B.update_input_task_set({{maneuver_{t}_day{i+1}_A}}, BaseTaskDependency.FF)")
        if j>0:
            prev_t = shifts[j-1]
            exec(f"maneuver_{t}_day{i+1}_A.update_input_task_set({{maneuver_{prev_t}_day{i+1}_A}})")
    if i != 0:
        exec(f"maneuver_04_1_day{i+1}_A.update_input_task_set({{maneuver_80_2_day{i}_A}})")

    exec(f"eng_day{i+1} = project.create_workflow('Engine Day{i+1}')")
    shifts = ['04_1', '48_1', '80_1', '04_2', '48_2', '80_2']
    for j, t in enumerate(shifts):
        exec(f"eng_{t}_day{i+1}_A = eng_day{i+1}.create_task('Engine {t} Day{i+1} A', need_facility=True, default_work_amount=240)")
        exec(f"eng_{t}_day{i+1}_B = eng_day{i+1}.create_task('Engine {t} Day{i+1} B', need_facility=True, default_work_amount=240)")
        exec(f"eng_sys.update_targeted_task_set({{eng_{t}_day{i+1}_A, eng_{t}_day{i+1}_B}})")
        exec(f"eng_room.update_targeted_task_set({{eng_{t}_day{i+1}_A, eng_{t}_day{i+1}_B}})")
        exec(f"tasks_engine_t1.append(eng_{t}_day{i+1}_A,)")
        exec(f"tasks_engine_t2.append(eng_{t}_day{i+1}_B,)")
        exec(f"eng_console_A.workamount_skill_mean_map.update({{eng_{t}_day{i+1}_A.name : 1.0,}})")
        exec(f"eng_console_B.workamount_skill_mean_map.update({{eng_{t}_day{i+1}_B.name : 1.0,}})")
        exec(f"eng_{t}_day{i+1}_B.update_input_task_set({{eng_{t}_day{i+1}_A}}, BaseTaskDependency.SS)")
        # exec(f"eng_{t}_day{i+1}_B.update_input_task_set({{eng_{t}_day{i+1}_A}}, BaseTaskDependency.FF)")
        if j>0:
            prev_t = shifts[j-1]
            exec(f"eng_{t}_day{i+1}_A.update_input_task_set({{eng_{prev_t}_day{i+1}_A}})")
    if i != 0:
        exec(f"eng_04_1_day{i+1}_A.update_input_task_set({{eng_80_2_day{i}_A}})")

    exec(f"elec_day{i+1} = project.create_workflow('Electric Day{i+1}')")
    shifts = ['04_1', '48_1', '80_1', '04_2', '48_2', '80_2']
    for j, t in enumerate(shifts):
        exec(f"elec_{t}_day{i+1}_A = elec_day{i+1}.create_task('Electric {t} Day{i+1} A', need_facility=True, default_work_amount=240)")
        exec(f"elec_sys.update_targeted_task_set({{elec_{t}_day{i+1}_A,}})")
        exec(f"elec_room.update_targeted_task_set({{elec_{t}_day{i+1}_A,}})")
        exec(f"tasks_elec_t1.append(elec_{t}_day{i+1}_A,)")
        exec(f"elec_console_A.workamount_skill_mean_map.update({{elec_{t}_day{i+1}_A.name : 1.0,}})")
        if j>0:
            prev_t = shifts[j-1]
            exec(f"elec_{t}_day{i+1}_A.update_input_task_set({{elec_{prev_t}_day{i+1}_A}})")
    if i != 0:
        exec(f"elec_04_1_day{i+1}_A.update_input_task_set({{elec_80_2_day{i}_A}})")
````

````python
for i in range(number_of_ur8k_survey):
    # 指示component・workflowの作成（仮）
    exec(f"ur8k_{i+1}_instruction_sys = ur8k.create_component('Urashima #{i+1} instruction system')")
    exec(f"ur8k_{i+1}_instruction = project.create_workflow('Instruction of Urashima #{i+1}')")


    # componentの作成
    exec(f"ur8k_{i+1}_launch_prep = ur8k.create_component('Urashima #{i+1} Launch Preparation')")
    

    # ---
    # workflow（準備）の作成
    exec(f"ur8k_{i+1}_Lpr = project.create_workflow('Preparing Urashima #{i+1} L')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Lpr = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Lpr', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lpr}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Lpr,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lpr}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Lpr.name: 1.0}})")

    # 格納庫　運航員task
    for t in "ABCDE":
        exec(f"ur8k_{i+1}_task_HA_{t}_Lpr = ur8k_{i+1}_Lpr.create_task('Task Hangar Urashima #{i+1} {t} Lpr', need_facility=True, default_work_amount=90)")

        exec(f"ur8k_{i+1}_launch_prep.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lpr}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lpr}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Lpr,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Lpr.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Lpr.update_input_task_set({{ur8k_{i+1}_task_HA_A_Lpr}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_HA_A_Lpr.update_input_task_set({{ur8k_{i+1}_task_instruction_Lpr}}, BaseTaskDependency.SS)")
    for t in "ABCDE":
        exec(f"ur8k_{i+1}_task_instruction_Lpr.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Lpr}}, BaseTaskDependency.FF)")
    # ---
  

    # ---
    # componentの作成
    exec(f"ur8k_{i+1}_launch = ur8k.create_component('Urashima #{i+1} Launch')")


    # ---
    # workflow（移動）の作成
    exec(f"ur8k_{i+1}_Lpo = project.create_workflow('Positioning Urashima #{i+1} L')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Lpo = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Lpo', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lpo}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Lpo,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lpo}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Lpo.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Lpo = ur8k_{i+1}_Lpo.create_task('Task C2room Urashima #{i+1} {t} Lpo', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Lpo}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Lpo}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Lpo,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Lpo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Lpo.update_input_task_set({{ur8k_{i+1}_task_C2_A_Lpo}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Lpo = ur8k_{i+1}_Lpo.create_task('Task Hangar Urashima #{i+1} {t} Lpo', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lpo}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lpo}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Lpo,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Lpo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Lpo.update_input_task_set({{ur8k_{i+1}_task_HA_A_Lpo}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Lpo = ur8k_{i+1}_Lpo.create_task('Task Bridge Urashima #{i+1} {t} Lpo', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Lpo}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Lpo}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Lpo,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Lpo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Lpo.update_input_task_set({{ur8k_{i+1}_task_BR_A_Lpo}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Lpo = ur8k_{i+1}_Lpo.create_task('Task Deck Urashima #{i+1} {t} Lpo', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lpo}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lpo}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Lpo,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Lpo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Lpo.update_input_task_set({{ur8k_{i+1}_task_DE_A_Lpo}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Lpo = ur8k_{i+1}_Lpo.create_task('Task Deck Urashima #{i+1} {t} Lpo', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lpo}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lpo}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Lpo,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Lpo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Lpo.update_input_task_set({{ur8k_{i+1}_task_DE_A_Lpo}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Lpo.update_input_task_set({{ur8k_{i+1}_task_instruction_Lpo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Lpo.update_input_task_set({{ur8k_{i+1}_task_instruction_Lpo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Lpo.update_input_task_set({{ur8k_{i+1}_task_instruction_Lpo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Lpo.update_input_task_set({{ur8k_{i+1}_task_instruction_Lpo}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Lpo.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Lpo}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Lpo.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Lpo}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Lpo.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Lpo}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Lpo.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Lpo}}, BaseTaskDependency.FF)")
    # ---

    
    # ---
    # workflow（玉掛け）の作成
    exec(f"ur8k_{i+1}_Lri = project.create_workflow('Rigging Urashima #{i+1} L')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Lri = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Lri', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lri}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Lri,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lri}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Lri.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Lri = ur8k_{i+1}_Lri.create_task('Task C2room Urashima #{i+1} {t} Lri', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Lri}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Lri}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Lri,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Lri.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Lri.update_input_task_set({{ur8k_{i+1}_task_C2_A_Lri}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Lri = ur8k_{i+1}_Lri.create_task('Task Hangar Urashima #{i+1} {t} Lri', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lri}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lri}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Lri,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Lri.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Lri.update_input_task_set({{ur8k_{i+1}_task_HA_A_Lri}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Lri = ur8k_{i+1}_Lri.create_task('Task Bridge Urashima #{i+1} {t} Lri', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Lri}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Lri}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Lri,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Lri.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Lri.update_input_task_set({{ur8k_{i+1}_task_BR_A_Lri}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Lri = ur8k_{i+1}_Lri.create_task('Task Deck Urashima #{i+1} {t} Lri', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lri}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lri}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Lri,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Lri.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Lri.update_input_task_set({{ur8k_{i+1}_task_DE_A_Lri}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Lri = ur8k_{i+1}_Lri.create_task('Task Deck Urashima #{i+1} {t} Lri', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lri}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lri}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Lri,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Lri.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Lri.update_input_task_set({{ur8k_{i+1}_task_DE_A_Lri}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Lri.update_input_task_set({{ur8k_{i+1}_task_instruction_Lri}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Lri.update_input_task_set({{ur8k_{i+1}_task_instruction_Lri}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Lri.update_input_task_set({{ur8k_{i+1}_task_instruction_Lri}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Lri.update_input_task_set({{ur8k_{i+1}_task_instruction_Lri}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Lri.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Lri}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Lri.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Lri}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Lri.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Lri}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Lri.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Lri}}, BaseTaskDependency.FF)")
    # ---


    # ---
    # workflow（フック固定）の作成
    exec(f"ur8k_{i+1}_Lho = project.create_workflow('Hooking Urashima #{i+1} L')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Lho = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Lho', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lho}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Lho,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lho}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Lho.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Lho = ur8k_{i+1}_Lho.create_task('Task C2room Urashima #{i+1} {t} Lho', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Lho}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Lho}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Lho,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Lho.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Lho.update_input_task_set({{ur8k_{i+1}_task_C2_A_Lho}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Lho = ur8k_{i+1}_Lho.create_task('Task Hangar Urashima #{i+1} {t} Lho', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lho}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lho}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Lho,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Lho.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Lho.update_input_task_set({{ur8k_{i+1}_task_HA_A_Lho}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Lho = ur8k_{i+1}_Lho.create_task('Task Bridge Urashima #{i+1} {t} Lho', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Lho}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Lho}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Lho,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Lho.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Lho.update_input_task_set({{ur8k_{i+1}_task_BR_A_Lho}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Lho = ur8k_{i+1}_Lho.create_task('Task Deck Urashima #{i+1} {t} Lho', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lho}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lho}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Lho,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Lho.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Lho.update_input_task_set({{ur8k_{i+1}_task_DE_A_Lho}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Lho = ur8k_{i+1}_Lho.create_task('Task Deck Urashima #{i+1} {t} Lho', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lho}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lho}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Lho,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Lho.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Lho.update_input_task_set({{ur8k_{i+1}_task_DE_A_Lho}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Lho.update_input_task_set({{ur8k_{i+1}_task_instruction_Lho}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Lho.update_input_task_set({{ur8k_{i+1}_task_instruction_Lho}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Lho.update_input_task_set({{ur8k_{i+1}_task_instruction_Lho}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Lho.update_input_task_set({{ur8k_{i+1}_task_instruction_Lho}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Lho.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Lho}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Lho.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Lho}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Lho.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Lho}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Lho.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Lho}}, BaseTaskDependency.FF)")
    # ---


    # ---
    # workflow（吊り上げ）の作成
    exec(f"ur8k_{i+1}_Lli = project.create_workflow('Lifting Urashima #{i+1} L')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Lli = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Lli', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lli}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Lli,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lli}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Lli.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Lli = ur8k_{i+1}_Lli.create_task('Task C2room Urashima #{i+1} {t} Lli', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Lli}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Lli}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Lli,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Lli.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Lli.update_input_task_set({{ur8k_{i+1}_task_C2_A_Lli}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Lli = ur8k_{i+1}_Lli.create_task('Task Hangar Urashima #{i+1} {t} Lli', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lli}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lli}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Lli,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Lli.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Lli.update_input_task_set({{ur8k_{i+1}_task_HA_A_Lli}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Lli = ur8k_{i+1}_Lli.create_task('Task Bridge Urashima #{i+1} {t} Lli', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Lli}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Lli}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Lli,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Lli.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Lli.update_input_task_set({{ur8k_{i+1}_task_BR_A_Lli}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Lli = ur8k_{i+1}_Lli.create_task('Task Deck Urashima #{i+1} {t} Lli', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lli}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lli}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Lli,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Lli.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Lli.update_input_task_set({{ur8k_{i+1}_task_DE_A_Lli}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Lli = ur8k_{i+1}_Lli.create_task('Task Deck Urashima #{i+1} {t} Lli', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lli}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lli}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Lli,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Lli.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Lli.update_input_task_set({{ur8k_{i+1}_task_DE_A_Lli}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Lli.update_input_task_set({{ur8k_{i+1}_task_instruction_Lli}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Lli.update_input_task_set({{ur8k_{i+1}_task_instruction_Lli}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Lli.update_input_task_set({{ur8k_{i+1}_task_instruction_Lli}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Lli.update_input_task_set({{ur8k_{i+1}_task_instruction_Lli}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Lli.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Lli}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Lli.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Lli}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Lli.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Lli}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Lli.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Lli}}, BaseTaskDependency.FF)")
    # ---

    
    # ---
    # workflow（船上から洋上へ）の作成
    exec(f"ur8k_{i+1}_Lsw = project.create_workflow('Swing Out Urashima #{i+1} L')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Lsw = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Lsw', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lsw}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Lsw,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lsw}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Lsw.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Lsw = ur8k_{i+1}_Lsw.create_task('Task C2room Urashima #{i+1} {t} Lsw', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Lsw}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Lsw}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Lsw,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Lsw.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Lsw.update_input_task_set({{ur8k_{i+1}_task_C2_A_Lsw}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Lsw = ur8k_{i+1}_Lsw.create_task('Task Hangar Urashima #{i+1} {t} Lsw', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lsw}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lsw}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Lsw,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Lsw.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Lsw.update_input_task_set({{ur8k_{i+1}_task_HA_A_Lsw}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Lsw = ur8k_{i+1}_Lsw.create_task('Task Bridge Urashima #{i+1} {t} Lsw', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Lsw}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Lsw}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Lsw,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Lsw.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Lsw.update_input_task_set({{ur8k_{i+1}_task_BR_A_Lsw}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Lsw = ur8k_{i+1}_Lsw.create_task('Task Deck Urashima #{i+1} {t} Lsw', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lsw}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lsw}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Lsw,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Lsw.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Lsw.update_input_task_set({{ur8k_{i+1}_task_DE_A_Lsw}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Lsw = ur8k_{i+1}_Lsw.create_task('Task Deck Urashima #{i+1} {t} Lsw', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lsw}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lsw}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Lsw,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Lsw.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Lsw.update_input_task_set({{ur8k_{i+1}_task_DE_A_Lsw}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Lsw.update_input_task_set({{ur8k_{i+1}_task_instruction_Lsw}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Lsw.update_input_task_set({{ur8k_{i+1}_task_instruction_Lsw}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Lsw.update_input_task_set({{ur8k_{i+1}_task_instruction_Lsw}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Lsw.update_input_task_set({{ur8k_{i+1}_task_instruction_Lsw}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Lsw.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Lsw}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Lsw.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Lsw}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Lsw.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Lsw}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Lsw.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Lsw}}, BaseTaskDependency.FF)")
    # ---

    
    # ---
    # workflow（降下）の作成
    exec(f"ur8k_{i+1}_Llo = project.create_workflow('Lowering Urashima #{i+1} L')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Llo = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Llo', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Llo}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Llo,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Llo}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Llo.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Llo = ur8k_{i+1}_Llo.create_task('Task C2room Urashima #{i+1} {t} Llo', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Llo}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Llo}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Llo,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Llo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Llo.update_input_task_set({{ur8k_{i+1}_task_C2_A_Llo}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Llo = ur8k_{i+1}_Llo.create_task('Task Hangar Urashima #{i+1} {t} Llo', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Llo}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Llo}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Llo,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Llo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Llo.update_input_task_set({{ur8k_{i+1}_task_HA_A_Llo}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Llo = ur8k_{i+1}_Llo.create_task('Task Bridge Urashima #{i+1} {t} Llo', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Llo}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Llo}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Llo,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Llo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Llo.update_input_task_set({{ur8k_{i+1}_task_BR_A_Llo}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Llo = ur8k_{i+1}_Llo.create_task('Task Deck Urashima #{i+1} {t} Llo', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Llo}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Llo}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Llo,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Llo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Llo.update_input_task_set({{ur8k_{i+1}_task_DE_A_Llo}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Llo = ur8k_{i+1}_Llo.create_task('Task Deck Urashima #{i+1} {t} Llo', need_facility=True, default_work_amount=3)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Llo}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Llo}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Llo,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Llo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Llo.update_input_task_set({{ur8k_{i+1}_task_DE_A_Llo}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Llo.update_input_task_set({{ur8k_{i+1}_task_instruction_Llo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Llo.update_input_task_set({{ur8k_{i+1}_task_instruction_Llo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Llo.update_input_task_set({{ur8k_{i+1}_task_instruction_Llo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Llo.update_input_task_set({{ur8k_{i+1}_task_instruction_Llo}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Llo.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Llo}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Llo.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Llo}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Llo.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Llo}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Llo.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Llo}}, BaseTaskDependency.FF)")
    # ---

    
    # ---
    # workflow（フック取外し）の作成
    exec(f"ur8k_{i+1}_Lre = project.create_workflow('Releasing Urashima #{i+1} L')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Lre = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Lre', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lre}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Lre,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Lre}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Lre.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Lre = ur8k_{i+1}_Lre.create_task('Task C2room Urashima #{i+1} {t} Lre', need_facility=True, default_work_amount=1)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Lre}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Lre}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Lre,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Lre.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Lre.update_input_task_set({{ur8k_{i+1}_task_C2_A_Lre}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Lre = ur8k_{i+1}_Lre.create_task('Task Hangar Urashima #{i+1} {t} Lre', need_facility=True, default_work_amount=1)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lre}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Lre}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Lre,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Lre.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Lre.update_input_task_set({{ur8k_{i+1}_task_HA_A_Lre}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Lre = ur8k_{i+1}_Lre.create_task('Task Bridge Urashima #{i+1} {t} Lre', need_facility=True, default_work_amount=1)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Lre}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Lre}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Lre,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Lre.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Lre.update_input_task_set({{ur8k_{i+1}_task_BR_A_Lre}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Lre = ur8k_{i+1}_Lre.create_task('Task Deck Urashima #{i+1} {t} Lre', need_facility=True, default_work_amount=1)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lre}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lre}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Lre,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Lre.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Lre.update_input_task_set({{ur8k_{i+1}_task_DE_A_Lre}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Lre = ur8k_{i+1}_Lre.create_task('Task Deck Urashima #{i+1} {t} Lre', need_facility=True, default_work_amount=1)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lre}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Lre}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Lre,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Lre.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Lre.update_input_task_set({{ur8k_{i+1}_task_DE_A_Lre}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Lre.update_input_task_set({{ur8k_{i+1}_task_instruction_Lre}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Lre.update_input_task_set({{ur8k_{i+1}_task_instruction_Lre}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Lre.update_input_task_set({{ur8k_{i+1}_task_instruction_Lre}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Lre.update_input_task_set({{ur8k_{i+1}_task_instruction_Lre}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Lre.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Lre}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Lre.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Lre}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Lre.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Lre}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Lre.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Lre}}, BaseTaskDependency.FF)")
    # ---

    
    # ---
    # workflow（潜入）の作成
    exec(f"ur8k_{i+1}_Ldi = project.create_workflow('Diving Urashima #{i+1} L')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Ldi = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Ldi', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Ldi}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Ldi,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Ldi}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Ldi.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Ldi = ur8k_{i+1}_Ldi.create_task('Task C2room Urashima #{i+1} {t} Ldi', need_facility=True, default_work_amount=1)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Ldi}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Ldi}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Ldi,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Ldi.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Ldi.update_input_task_set({{ur8k_{i+1}_task_C2_A_Ldi}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Ldi = ur8k_{i+1}_Ldi.create_task('Task Hangar Urashima #{i+1} {t} Ldi', need_facility=True, default_work_amount=1)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Ldi}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Ldi}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Ldi,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Ldi.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Ldi.update_input_task_set({{ur8k_{i+1}_task_HA_A_Ldi}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Ldi = ur8k_{i+1}_Ldi.create_task('Task Bridge Urashima #{i+1} {t} Ldi', need_facility=True, default_work_amount=1)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Ldi}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Ldi}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Ldi,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Ldi.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Ldi.update_input_task_set({{ur8k_{i+1}_task_BR_A_Ldi}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Ldi = ur8k_{i+1}_Ldi.create_task('Task Deck Urashima #{i+1} {t} Ldi', need_facility=True, default_work_amount=1)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Ldi}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Ldi}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Ldi,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Ldi.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Ldi.update_input_task_set({{ur8k_{i+1}_task_DE_A_Ldi}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Ldi = ur8k_{i+1}_Ldi.create_task('Task Deck Urashima #{i+1} {t} Ldi', need_facility=True, default_work_amount=1)")

        exec(f"ur8k_{i+1}_launch.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Ldi}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Ldi}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Ldi,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Ldi.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Ldi.update_input_task_set({{ur8k_{i+1}_task_DE_A_Ldi}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Ldi.update_input_task_set({{ur8k_{i+1}_task_instruction_Ldi}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Ldi.update_input_task_set({{ur8k_{i+1}_task_instruction_Ldi}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Ldi.update_input_task_set({{ur8k_{i+1}_task_instruction_Ldi}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Ldi.update_input_task_set({{ur8k_{i+1}_task_instruction_Ldi}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Ldi.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Ldi}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Ldi.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Ldi}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Ldi.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Ldi}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Ldi.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Ldi}}, BaseTaskDependency.FF)")
    # ---


    # ---
    # 異なるworkflow間のtask依存関係
    exec(f"ur8k_{i+1}_task_instruction_Lpo.update_input_task_set({{ur8k_{i+1}_task_instruction_Lpr}})")
    exec(f"ur8k_{i+1}_task_instruction_Lri.update_input_task_set({{ur8k_{i+1}_task_instruction_Lpo}})")
    exec(f"ur8k_{i+1}_task_instruction_Lho.update_input_task_set({{ur8k_{i+1}_task_instruction_Lri}})")
    exec(f"ur8k_{i+1}_task_instruction_Lli.update_input_task_set({{ur8k_{i+1}_task_instruction_Lho}})")
    exec(f"ur8k_{i+1}_task_instruction_Lsw.update_input_task_set({{ur8k_{i+1}_task_instruction_Lli}})")
    exec(f"ur8k_{i+1}_task_instruction_Llo.update_input_task_set({{ur8k_{i+1}_task_instruction_Lsw}})")
    exec(f"ur8k_{i+1}_task_instruction_Lre.update_input_task_set({{ur8k_{i+1}_task_instruction_Llo}})")
    exec(f"ur8k_{i+1}_task_instruction_Ldi.update_input_task_set({{ur8k_{i+1}_task_instruction_Lre}})")
    # ---


    
    # ---
    # componentの作成
    exec(f"ur8k_{i+1}_survey = ur8k.create_component('Urashima #{i+1} Survey')")


    # workflow（探査）の作成
    exec(f"ur8k_{i+1}_survey1 = project.create_workflow('Survey Urashima #{i+1}')")

    # taskの作成
    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_control_{t}_Ss1 = ur8k_{i+1}_survey1.create_task('Control Urashima #{i+1} {t} Ss1', need_facility=True, default_work_amount=240)")

        exec(f"ur8k_{i+1}_survey.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss1}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss1}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_control_{t}_Ss1,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_control_{t}_Ss1.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_control_{t}_Ss1.update_input_task_set({{ur8k_{i+1}_control_A_Ss1}}, BaseTaskDependency.SS)")
            # exec(f"ur8k_{i+1}_control_{t}_Ss1.update_input_task_set({{ur8k_{i+1}_control_A_Ss1}}, BaseTaskDependency.FF)")

    for t in "AB":
        exec(f"ur8k_{i+1}_control_{t}_Ss2 = ur8k_{i+1}_survey1.create_task('Control Urashima #{i+1} {t} Ss2', need_facility=True, default_work_amount=240)")

        exec(f"ur8k_{i+1}_survey.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss2}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss2}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_control_{t}_Ss2,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_control_{t}_Ss2.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_control_{t}_Ss2.update_input_task_set({{ur8k_{i+1}_control_A_Ss2}}, BaseTaskDependency.SS)")
            # exec(f"ur8k_{i+1}_control_{t}_Ss2.update_input_task_set({{ur8k_{i+1}_control_A_Ss2}}, BaseTaskDependency.FF)")
    
    for t in "AB":
        exec(f"ur8k_{i+1}_control_{t}_Ss3 = ur8k_{i+1}_survey1.create_task('Control Urashima #{i+1} {t} Ss3', need_facility=True, default_work_amount=240)")

        exec(f"ur8k_{i+1}_survey.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss3}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss3}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_control_{t}_Ss3,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_control_{t}_Ss3.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_control_{t}_Ss3.update_input_task_set({{ur8k_{i+1}_control_A_Ss3}}, BaseTaskDependency.SS)")
            # exec(f"ur8k_{i+1}_control_{t}_Ss3.update_input_task_set({{ur8k_{i+1}_control_A_Ss3}}, BaseTaskDependency.FF)")

    for t in "AB":
        exec(f"ur8k_{i+1}_control_{t}_Ss4 = ur8k_{i+1}_survey1.create_task('Control Urashima #{i+1} {t} Ss4', need_facility=True, default_work_amount=240)")

        exec(f"ur8k_{i+1}_survey.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss4}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss4}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_control_{t}_Ss4,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_control_{t}_Ss4.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_control_{t}_Ss4.update_input_task_set({{ur8k_{i+1}_control_A_Ss4}}, BaseTaskDependency.SS)")
            # exec(f"ur8k_{i+1}_control_{t}_Ss4.update_input_task_set({{ur8k_{i+1}_control_A_Ss4}}, BaseTaskDependency.FF)")
    
    for t in "AB":
        exec(f"ur8k_{i+1}_control_{t}_Ss5 = ur8k_{i+1}_survey1.create_task('Control Urashima #{i+1} {t} Ss5', need_facility=True, default_work_amount=240)")

        exec(f"ur8k_{i+1}_survey.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss5}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss5}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_control_{t}_Ss5,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_control_{t}_Ss5.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_control_{t}_Ss5.update_input_task_set({{ur8k_{i+1}_control_A_Ss5}}, BaseTaskDependency.SS)")
            # exec(f"ur8k_{i+1}_control_{t}_Ss5.update_input_task_set({{ur8k_{i+1}_control_A_Ss5}}, BaseTaskDependency.FF)")

    for t in "AB":
        exec(f"ur8k_{i+1}_control_{t}_Ss6 = ur8k_{i+1}_survey1.create_task('Control Urashima #{i+1} {t} Ss6', need_facility=True, default_work_amount=240)")

        exec(f"ur8k_{i+1}_survey.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss6}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss6}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_control_{t}_Ss6,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_control_{t}_Ss6.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_control_{t}_Ss6.update_input_task_set({{ur8k_{i+1}_control_A_Ss6}}, BaseTaskDependency.SS)")
            # exec(f"ur8k_{i+1}_control_{t}_Ss6.update_input_task_set({{ur8k_{i+1}_control_A_Ss6}}, BaseTaskDependency.FF)")

    for t in "AB":
        exec(f"ur8k_{i+1}_control_{t}_Ss7 = ur8k_{i+1}_survey1.create_task('Control Urashima #{i+1} {t} Ss7', need_facility=True, default_work_amount=240)")

        exec(f"ur8k_{i+1}_survey.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss7}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_control_{t}_Ss7}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_control_{t}_Ss7,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_control_{t}_Ss7.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_control_{t}_Ss7.update_input_task_set({{ur8k_{i+1}_control_A_Ss7}}, BaseTaskDependency.SS)")
            # exec(f"ur8k_{i+1}_control_{t}_Ss7.update_input_task_set({{ur8k_{i+1}_control_A_Ss7}}, BaseTaskDependency.FF)")
    
    # 異なるworkflow間のtask依存関係
    exec(f"ur8k_{i+1}_control_A_Ss1.update_input_task_set({{ur8k_{i+1}_task_C2_A_Ldi}})")
    exec(f"ur8k_{i+1}_control_A_Ss2.update_input_task_set({{ur8k_{i+1}_control_A_Ss1}})")
    exec(f"ur8k_{i+1}_control_A_Ss3.update_input_task_set({{ur8k_{i+1}_control_A_Ss2}})")
    exec(f"ur8k_{i+1}_control_A_Ss4.update_input_task_set({{ur8k_{i+1}_control_A_Ss3}})")
    exec(f"ur8k_{i+1}_control_A_Ss5.update_input_task_set({{ur8k_{i+1}_control_A_Ss4}})")
    exec(f"ur8k_{i+1}_control_A_Ss6.update_input_task_set({{ur8k_{i+1}_control_A_Ss5}})")
    exec(f"ur8k_{i+1}_control_A_Ss7.update_input_task_set({{ur8k_{i+1}_control_A_Ss6}})")
    # ---



    # ---
    # componentの作成
    exec(f"ur8k_{i+1}_recover = ur8k.create_component('Urashima #{i+1} Recover')")
    # ---


    # ---
    # workflow（浮上）の作成
    exec(f"ur8k_{i+1}_Rsu = project.create_workflow('Surfacing Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rsu = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rsu', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rsu}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rsu,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rsu}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rsu.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Rsu = ur8k_{i+1}_Rsu.create_task('Task C2room Urashima #{i+1} {t} Rsu', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rsu}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rsu}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Rsu,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Rsu.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Rsu.update_input_task_set({{ur8k_{i+1}_task_C2_A_Rsu}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rsu = ur8k_{i+1}_Rsu.create_task('Task Hangar Urashima #{i+1} {t} Rsu', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rsu}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rsu}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rsu,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rsu.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rsu.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rsu}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Rsu = ur8k_{i+1}_Rsu.create_task('Task Bridge Urashima #{i+1} {t} Rsu', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rsu}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rsu}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Rsu,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Rsu.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Rsu.update_input_task_set({{ur8k_{i+1}_task_BR_A_Rsu}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rsu = ur8k_{i+1}_Rsu.create_task('Task Deck Urashima #{i+1} {t} Rsu', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rsu}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rsu}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Rsu,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rsu.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rsu.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rsu}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rsu = ur8k_{i+1}_Rsu.create_task('Task Deck Urashima #{i+1} {t} Rsu', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rsu}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rsu}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Rsu,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rsu.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rsu.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rsu}}, BaseTaskDependency.SS)")

    # 作業艇　船員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rsu = ur8k_{i+1}_Rsu.create_task('Task Boat Urashima #{i+1} {t} Rsu', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rsu}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rsu}})")
        exec(f"tasks_boat_t1.append(ur8k_{i+1}_task_BO_{t}_Rsu,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rsu.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rsu.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rsu}}, BaseTaskDependency.SS)")
    for t in "CD":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rsu = ur8k_{i+1}_Rsu.create_task('Task Boat Urashima #{i+1} {t} Rsu', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rsu}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rsu}})")
        exec(f"tasks_boat_t2.append(ur8k_{i+1}_task_BO_{t}_Rsu,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rsu.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rsu.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rsu}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Rsu.update_input_task_set({{ur8k_{i+1}_task_instruction_Rsu}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Rsu.update_input_task_set({{ur8k_{i+1}_task_instruction_Rsu}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Rsu.update_input_task_set({{ur8k_{i+1}_task_instruction_Rsu}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Rsu.update_input_task_set({{ur8k_{i+1}_task_instruction_Rsu}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BO_A_Rsu.update_input_task_set({{ur8k_{i+1}_task_instruction_Rsu}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Rsu.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Rsu}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rsu.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rsu}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Rsu.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Rsu}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Rsu.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Rsu}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rsu.update_input_task_set({{ur8k_{i+1}_task_BO_{t}_Rsu}}, BaseTaskDependency.FF)")
    # ---

    
    # ---
    # workflow（先取り）の作成
    exec(f"ur8k_{i+1}_Rca = project.create_workflow('Caputuring Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rca = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rca', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rca}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rca,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rca}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rca.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Rca = ur8k_{i+1}_Rca.create_task('Task C2room Urashima #{i+1} {t} Rca', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rca}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rca}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Rca,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Rca.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Rca.update_input_task_set({{ur8k_{i+1}_task_C2_A_Rca}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rca = ur8k_{i+1}_Rca.create_task('Task Hangar Urashima #{i+1} {t} Rca', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rca}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rca}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rca,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rca.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rca.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rca}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Rca = ur8k_{i+1}_Rca.create_task('Task Bridge Urashima #{i+1} {t} Rca', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rca}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rca}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Rca,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Rca.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Rca.update_input_task_set({{ur8k_{i+1}_task_BR_A_Rca}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rca = ur8k_{i+1}_Rca.create_task('Task Deck Urashima #{i+1} {t} Rca', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rca}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rca}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Rca,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rca.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rca.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rca}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rca = ur8k_{i+1}_Rca.create_task('Task Deck Urashima #{i+1} {t} Rca', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rca}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rca}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Rca,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rca.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rca.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rca}}, BaseTaskDependency.SS)")

    # 作業艇　船員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rca = ur8k_{i+1}_Rca.create_task('Task Boat Urashima #{i+1} {t} Rca', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rca}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rca}})")
        exec(f"tasks_boat_t1.append(ur8k_{i+1}_task_BO_{t}_Rca,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rca.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rca.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rca}}, BaseTaskDependency.SS)")
    for t in "CD":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rca = ur8k_{i+1}_Rca.create_task('Task Boat Urashima #{i+1} {t} Rca', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rca}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rca}})")
        exec(f"tasks_boat_t2.append(ur8k_{i+1}_task_BO_{t}_Rca,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rca.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rca.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rca}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Rca.update_input_task_set({{ur8k_{i+1}_task_instruction_Rca}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Rca.update_input_task_set({{ur8k_{i+1}_task_instruction_Rca}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Rca.update_input_task_set({{ur8k_{i+1}_task_instruction_Rca}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Rca.update_input_task_set({{ur8k_{i+1}_task_instruction_Rca}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BO_A_Rca.update_input_task_set({{ur8k_{i+1}_task_instruction_Rca}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Rca.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Rca}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rca.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rca}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Rca.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Rca}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Rca.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Rca}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rca.update_input_task_set({{ur8k_{i+1}_task_BO_{t}_Rca}}, BaseTaskDependency.FF)")
    # ---
    

    # ---
    # workflow（吊り位置に移動）の作成
    exec(f"ur8k_{i+1}_Rdr = project.create_workflow('Dragging Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rdr = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rdr', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rdr}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rdr,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rdr}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rdr.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Rdr = ur8k_{i+1}_Rdr.create_task('Task C2room Urashima #{i+1} {t} Rdr', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rdr}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rdr}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Rdr,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Rdr.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Rdr.update_input_task_set({{ur8k_{i+1}_task_C2_A_Rdr}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rdr = ur8k_{i+1}_Rdr.create_task('Task Hangar Urashima #{i+1} {t} Rdr', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rdr}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rdr}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rdr,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rdr.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rdr.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rdr}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Rdr = ur8k_{i+1}_Rdr.create_task('Task Bridge Urashima #{i+1} {t} Rdr', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rdr}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rdr}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Rdr,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Rdr.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Rdr.update_input_task_set({{ur8k_{i+1}_task_BR_A_Rdr}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rdr = ur8k_{i+1}_Rdr.create_task('Task Deck Urashima #{i+1} {t} Rdr', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rdr}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rdr}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Rdr,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rdr.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rdr.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rdr}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rdr = ur8k_{i+1}_Rdr.create_task('Task Deck Urashima #{i+1} {t} Rdr', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rdr}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rdr}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Rdr,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rdr.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rdr.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rdr}}, BaseTaskDependency.SS)")

    # 作業艇　船員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rdr = ur8k_{i+1}_Rdr.create_task('Task Boat Urashima #{i+1} {t} Rdr', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rdr}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rdr}})")
        exec(f"tasks_boat_t1.append(ur8k_{i+1}_task_BO_{t}_Rdr,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rdr.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rdr.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rdr}}, BaseTaskDependency.SS)")
    for t in "CD":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rdr = ur8k_{i+1}_Rdr.create_task('Task Boat Urashima #{i+1} {t} Rdr', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rdr}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rdr}})")
        exec(f"tasks_boat_t2.append(ur8k_{i+1}_task_BO_{t}_Rdr,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rdr.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rdr.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rdr}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Rdr.update_input_task_set({{ur8k_{i+1}_task_instruction_Rdr}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Rdr.update_input_task_set({{ur8k_{i+1}_task_instruction_Rdr}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Rdr.update_input_task_set({{ur8k_{i+1}_task_instruction_Rdr}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Rdr.update_input_task_set({{ur8k_{i+1}_task_instruction_Rdr}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BO_A_Rdr.update_input_task_set({{ur8k_{i+1}_task_instruction_Rdr}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Rdr.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Rdr}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rdr.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rdr}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Rdr.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Rdr}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Rdr.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Rdr}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rdr.update_input_task_set({{ur8k_{i+1}_task_BO_{t}_Rdr}}, BaseTaskDependency.FF)")
    # ---

    
    # ---
    # workflow（玉掛け）の作成
    exec(f"ur8k_{i+1}_Rri = project.create_workflow('Rigging Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rri = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rri', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rri}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rri,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rri}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rri.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Rri = ur8k_{i+1}_Rri.create_task('Task C2room Urashima #{i+1} {t} Rri', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rri}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rri}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Rri,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Rri.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Rri.update_input_task_set({{ur8k_{i+1}_task_C2_A_Rri}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rri = ur8k_{i+1}_Rri.create_task('Task Hangar Urashima #{i+1} {t} Rri', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rri}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rri}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rri,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rri.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rri.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rri}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Rri = ur8k_{i+1}_Rri.create_task('Task Bridge Urashima #{i+1} {t} Rri', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rri}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rri}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Rri,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Rri.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Rri.update_input_task_set({{ur8k_{i+1}_task_BR_A_Rri}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rri = ur8k_{i+1}_Rri.create_task('Task Deck Urashima #{i+1} {t} Rri', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rri}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rri}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Rri,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rri.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rri.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rri}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rri = ur8k_{i+1}_Rri.create_task('Task Deck Urashima #{i+1} {t} Rri', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rri}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rri}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Rri,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rri.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rri.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rri}}, BaseTaskDependency.SS)")

    # 作業艇　船員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rri = ur8k_{i+1}_Rri.create_task('Task Boat Urashima #{i+1} {t} Rri', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rri}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rri}})")
        exec(f"tasks_boat_t1.append(ur8k_{i+1}_task_BO_{t}_Rri,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rri.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rri.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rri}}, BaseTaskDependency.SS)")
    for t in "CD":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rri = ur8k_{i+1}_Rri.create_task('Task Boat Urashima #{i+1} {t} Rri', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rri}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rri}})")
        exec(f"tasks_boat_t2.append(ur8k_{i+1}_task_BO_{t}_Rri,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rri.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rri.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rri}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Rri.update_input_task_set({{ur8k_{i+1}_task_instruction_Rri}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Rri.update_input_task_set({{ur8k_{i+1}_task_instruction_Rri}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Rri.update_input_task_set({{ur8k_{i+1}_task_instruction_Rri}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Rri.update_input_task_set({{ur8k_{i+1}_task_instruction_Rri}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BO_A_Rri.update_input_task_set({{ur8k_{i+1}_task_instruction_Rri}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Rri.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Rri}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rri.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rri}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Rri.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Rri}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Rri.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Rri}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rri.update_input_task_set({{ur8k_{i+1}_task_BO_{t}_Rri}}, BaseTaskDependency.FF)")
    # ---

    
    # ---
    # workflow（フック固定）の作成
    exec(f"ur8k_{i+1}_Rho = project.create_workflow('Hooking Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rho = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rho', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rho}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rho,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rho}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rho.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Rho = ur8k_{i+1}_Rho.create_task('Task C2room Urashima #{i+1} {t} Rho', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rho}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rho}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Rho,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Rho.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Rho.update_input_task_set({{ur8k_{i+1}_task_C2_A_Rho}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rho = ur8k_{i+1}_Rho.create_task('Task Hangar Urashima #{i+1} {t} Rho', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rho}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rho}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rho,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rho.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rho.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rho}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Rho = ur8k_{i+1}_Rho.create_task('Task Bridge Urashima #{i+1} {t} Rho', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rho}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rho}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Rho,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Rho.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Rho.update_input_task_set({{ur8k_{i+1}_task_BR_A_Rho}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rho = ur8k_{i+1}_Rho.create_task('Task Deck Urashima #{i+1} {t} Rho', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rho}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rho}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Rho,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rho.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rho.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rho}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rho = ur8k_{i+1}_Rho.create_task('Task Deck Urashima #{i+1} {t} Rho', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rho}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rho}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Rho,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rho.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rho.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rho}}, BaseTaskDependency.SS)")

    # 作業艇　船員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rho = ur8k_{i+1}_Rho.create_task('Task Boat Urashima #{i+1} {t} Rho', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rho}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rho}})")
        exec(f"tasks_boat_t1.append(ur8k_{i+1}_task_BO_{t}_Rho,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rho.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rho.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rho}}, BaseTaskDependency.SS)")
    for t in "CD":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rho = ur8k_{i+1}_Rho.create_task('Task Boat Urashima #{i+1} {t} Rho', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rho}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rho}})")
        exec(f"tasks_boat_t2.append(ur8k_{i+1}_task_BO_{t}_Rho,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rho.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rho.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rho}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Rho.update_input_task_set({{ur8k_{i+1}_task_instruction_Rho}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Rho.update_input_task_set({{ur8k_{i+1}_task_instruction_Rho}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Rho.update_input_task_set({{ur8k_{i+1}_task_instruction_Rho}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Rho.update_input_task_set({{ur8k_{i+1}_task_instruction_Rho}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BO_A_Rho.update_input_task_set({{ur8k_{i+1}_task_instruction_Rho}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Rho.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Rho}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rho.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rho}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Rho.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Rho}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Rho.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Rho}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rho.update_input_task_set({{ur8k_{i+1}_task_BO_{t}_Rho}}, BaseTaskDependency.FF)")
    # ---

    
    # ---
    # workflow（吊り上げ）の作成
    exec(f"ur8k_{i+1}_Rli = project.create_workflow('Lifting Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rli = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rli', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rli}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rli,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rli}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rli.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Rli = ur8k_{i+1}_Rli.create_task('Task C2room Urashima #{i+1} {t} Rli', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rli}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rli}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Rli,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Rli.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Rli.update_input_task_set({{ur8k_{i+1}_task_C2_A_Rli}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rli = ur8k_{i+1}_Rli.create_task('Task Hangar Urashima #{i+1} {t} Rli', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rli}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rli}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rli,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rli.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rli.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rli}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Rli = ur8k_{i+1}_Rli.create_task('Task Bridge Urashima #{i+1} {t} Rli', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rli}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rli}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Rli,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Rli.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Rli.update_input_task_set({{ur8k_{i+1}_task_BR_A_Rli}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rli = ur8k_{i+1}_Rli.create_task('Task Deck Urashima #{i+1} {t} Rli', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rli}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rli}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Rli,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rli.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rli.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rli}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rli = ur8k_{i+1}_Rli.create_task('Task Deck Urashima #{i+1} {t} Rli', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rli}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rli}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Rli,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rli.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rli.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rli}}, BaseTaskDependency.SS)")

    # 作業艇　船員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rli = ur8k_{i+1}_Rli.create_task('Task Boat Urashima #{i+1} {t} Rli', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rli}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rli}})")
        exec(f"tasks_boat_t1.append(ur8k_{i+1}_task_BO_{t}_Rli,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rli.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rli.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rli}}, BaseTaskDependency.SS)")
    for t in "CD":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rli = ur8k_{i+1}_Rli.create_task('Task Boat Urashima #{i+1} {t} Rli', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rli}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rli}})")
        exec(f"tasks_boat_t2.append(ur8k_{i+1}_task_BO_{t}_Rli,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rli.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rli.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rli}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Rli.update_input_task_set({{ur8k_{i+1}_task_instruction_Rli}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Rli.update_input_task_set({{ur8k_{i+1}_task_instruction_Rli}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Rli.update_input_task_set({{ur8k_{i+1}_task_instruction_Rli}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Rli.update_input_task_set({{ur8k_{i+1}_task_instruction_Rli}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BO_A_Rli.update_input_task_set({{ur8k_{i+1}_task_instruction_Rli}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Rli.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Rli}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rli.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rli}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Rli.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Rli}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Rli.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Rli}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rli.update_input_task_set({{ur8k_{i+1}_task_BO_{t}_Rli}}, BaseTaskDependency.FF)")
    # ---


    # ---
    # workflow（洋上から船上へ）の作成
    exec(f"ur8k_{i+1}_Rsw = project.create_workflow('Swing In Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rsw = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rsw', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rsw}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rsw,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rsw}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rsw.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Rsw = ur8k_{i+1}_Rsw.create_task('Task C2room Urashima #{i+1} {t} Rsw', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rsw}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rsw}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Rsw,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Rsw.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Rsw.update_input_task_set({{ur8k_{i+1}_task_C2_A_Rsw}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rsw = ur8k_{i+1}_Rsw.create_task('Task Hangar Urashima #{i+1} {t} Rsw', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rsw}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rsw}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rsw,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rsw.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rsw.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rsw}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Rsw = ur8k_{i+1}_Rsw.create_task('Task Bridge Urashima #{i+1} {t} Rsw', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rsw}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rsw}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Rsw,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Rsw.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Rsw.update_input_task_set({{ur8k_{i+1}_task_BR_A_Rsw}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rsw = ur8k_{i+1}_Rsw.create_task('Task Deck Urashima #{i+1} {t} Rsw', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rsw}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rsw}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Rsw,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rsw.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rsw.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rsw}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rsw = ur8k_{i+1}_Rsw.create_task('Task Deck Urashima #{i+1} {t} Rsw', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rsw}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rsw}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Rsw,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rsw.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rsw.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rsw}}, BaseTaskDependency.SS)")

    # 作業艇　船員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rsw = ur8k_{i+1}_Rsw.create_task('Task Boat Urashima #{i+1} {t} Rsw', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rsw}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rsw}})")
        exec(f"tasks_boat_t1.append(ur8k_{i+1}_task_BO_{t}_Rsw,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rsw.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rsw.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rsw}}, BaseTaskDependency.SS)")
    for t in "CD":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rsw = ur8k_{i+1}_Rsw.create_task('Task Boat Urashima #{i+1} {t} Rsw', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rsw}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rsw}})")
        exec(f"tasks_boat_t2.append(ur8k_{i+1}_task_BO_{t}_Rsw,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rsw.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rsw.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rsw}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Rsw.update_input_task_set({{ur8k_{i+1}_task_instruction_Rsw}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Rsw.update_input_task_set({{ur8k_{i+1}_task_instruction_Rsw}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Rsw.update_input_task_set({{ur8k_{i+1}_task_instruction_Rsw}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Rsw.update_input_task_set({{ur8k_{i+1}_task_instruction_Rsw}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BO_A_Rsw.update_input_task_set({{ur8k_{i+1}_task_instruction_Rsw}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Rsw.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Rsw}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rsw.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rsw}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Rsw.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Rsw}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Rsw.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Rsw}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rsw.update_input_task_set({{ur8k_{i+1}_task_BO_{t}_Rsw}}, BaseTaskDependency.FF)")
    # ---

    
    # ---
    # workflow（降下）の作成
    exec(f"ur8k_{i+1}_Rlo = project.create_workflow('Lowering Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rlo = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rlo', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rlo}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rlo,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rlo}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rlo.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Rlo = ur8k_{i+1}_Rlo.create_task('Task C2room Urashima #{i+1} {t} Rlo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rlo}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rlo}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Rlo,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Rlo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Rlo.update_input_task_set({{ur8k_{i+1}_task_C2_A_Rlo}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rlo = ur8k_{i+1}_Rlo.create_task('Task Hangar Urashima #{i+1} {t} Rlo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rlo}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rlo}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rlo,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rlo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rlo.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rlo}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Rlo = ur8k_{i+1}_Rlo.create_task('Task Bridge Urashima #{i+1} {t} Rlo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rlo}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rlo}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Rlo,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Rlo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Rlo.update_input_task_set({{ur8k_{i+1}_task_BR_A_Rlo}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rlo = ur8k_{i+1}_Rlo.create_task('Task Deck Urashima #{i+1} {t} Rlo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rlo}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rlo}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Rlo,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rlo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rlo.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rlo}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rlo = ur8k_{i+1}_Rlo.create_task('Task Deck Urashima #{i+1} {t} Rlo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rlo}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rlo}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Rlo,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rlo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rlo.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rlo}}, BaseTaskDependency.SS)")

    # 作業艇　船員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rlo = ur8k_{i+1}_Rlo.create_task('Task Boat Urashima #{i+1} {t} Rlo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rlo}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rlo}})")
        exec(f"tasks_boat_t1.append(ur8k_{i+1}_task_BO_{t}_Rlo,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rlo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rlo.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rlo}}, BaseTaskDependency.SS)")
    for t in "CD":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rlo = ur8k_{i+1}_Rlo.create_task('Task Boat Urashima #{i+1} {t} Rlo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rlo}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rlo}})")
        exec(f"tasks_boat_t2.append(ur8k_{i+1}_task_BO_{t}_Rlo,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rlo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rlo.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rlo}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Rlo.update_input_task_set({{ur8k_{i+1}_task_instruction_Rlo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Rlo.update_input_task_set({{ur8k_{i+1}_task_instruction_Rlo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Rlo.update_input_task_set({{ur8k_{i+1}_task_instruction_Rlo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Rlo.update_input_task_set({{ur8k_{i+1}_task_instruction_Rlo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BO_A_Rlo.update_input_task_set({{ur8k_{i+1}_task_instruction_Rlo}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Rlo.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Rlo}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rlo.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rlo}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Rlo.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Rlo}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Rlo.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Rlo}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rlo.update_input_task_set({{ur8k_{i+1}_task_BO_{t}_Rlo}}, BaseTaskDependency.FF)")
    # ---

    
    # ---
    # workflow（着座）の作成
    exec(f"ur8k_{i+1}_Rla = project.create_workflow('Landing Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rla = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rla', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rla}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rla,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rla}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rla.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Rla = ur8k_{i+1}_Rla.create_task('Task C2room Urashima #{i+1} {t} Rla', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rla}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rla}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Rla,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Rla.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Rla.update_input_task_set({{ur8k_{i+1}_task_C2_A_Rla}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rla = ur8k_{i+1}_Rla.create_task('Task Hangar Urashima #{i+1} {t} Rla', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rla}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rla}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rla,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rla.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rla.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rla}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Rla = ur8k_{i+1}_Rla.create_task('Task Bridge Urashima #{i+1} {t} Rla', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rla}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rla}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Rla,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Rla.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Rla.update_input_task_set({{ur8k_{i+1}_task_BR_A_Rla}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rla = ur8k_{i+1}_Rla.create_task('Task Deck Urashima #{i+1} {t} Rla', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rla}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rla}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Rla,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rla.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rla.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rla}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rla = ur8k_{i+1}_Rla.create_task('Task Deck Urashima #{i+1} {t} Rla', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rla}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rla}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Rla,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rla.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rla.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rla}}, BaseTaskDependency.SS)")

    # 作業艇　船員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rla = ur8k_{i+1}_Rla.create_task('Task Boat Urashima #{i+1} {t} Rla', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rla}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rla}})")
        exec(f"tasks_boat_t1.append(ur8k_{i+1}_task_BO_{t}_Rla,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rla.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rla.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rla}}, BaseTaskDependency.SS)")
    for t in "CD":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rla = ur8k_{i+1}_Rla.create_task('Task Boat Urashima #{i+1} {t} Rla', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rla}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rla}})")
        exec(f"tasks_boat_t2.append(ur8k_{i+1}_task_BO_{t}_Rla,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rla.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rla.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rla}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Rla.update_input_task_set({{ur8k_{i+1}_task_instruction_Rla}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Rla.update_input_task_set({{ur8k_{i+1}_task_instruction_Rla}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Rla.update_input_task_set({{ur8k_{i+1}_task_instruction_Rla}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Rla.update_input_task_set({{ur8k_{i+1}_task_instruction_Rla}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BO_A_Rla.update_input_task_set({{ur8k_{i+1}_task_instruction_Rla}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Rla.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Rla}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rla.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rla}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Rla.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Rla}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Rla.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Rla}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rla.update_input_task_set({{ur8k_{i+1}_task_BO_{t}_Rla}}, BaseTaskDependency.FF)")
    # ---


    # ---
    # workflow（フック取り外し）の作成
    exec(f"ur8k_{i+1}_Rre = project.create_workflow('Release Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rre = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rre', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rre}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rre,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rre}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rre.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Rre = ur8k_{i+1}_Rre.create_task('Task C2room Urashima #{i+1} {t} Rre', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rre}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rre}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Rre,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Rre.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Rre.update_input_task_set({{ur8k_{i+1}_task_C2_A_Rre}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rre = ur8k_{i+1}_Rre.create_task('Task Hangar Urashima #{i+1} {t} Rre', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rre}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rre}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rre,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rre.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rre.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rre}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Rre = ur8k_{i+1}_Rre.create_task('Task Bridge Urashima #{i+1} {t} Rre', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rre}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rre}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Rre,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Rre.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Rre.update_input_task_set({{ur8k_{i+1}_task_BR_A_Rre}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rre = ur8k_{i+1}_Rre.create_task('Task Deck Urashima #{i+1} {t} Rre', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rre}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rre}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Rre,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rre.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rre.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rre}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rre = ur8k_{i+1}_Rre.create_task('Task Deck Urashima #{i+1} {t} Rre', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rre}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rre}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Rre,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rre.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rre.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rre}}, BaseTaskDependency.SS)")

    # 作業艇　船員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rre = ur8k_{i+1}_Rre.create_task('Task Boat Urashima #{i+1} {t} Rre', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rre}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rre}})")
        exec(f"tasks_boat_t1.append(ur8k_{i+1}_task_BO_{t}_Rre,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rre.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rre.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rre}}, BaseTaskDependency.SS)")
    for t in "CD":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rre = ur8k_{i+1}_Rre.create_task('Task Boat Urashima #{i+1} {t} Rre', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rre}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rre}})")
        exec(f"tasks_boat_t2.append(ur8k_{i+1}_task_BO_{t}_Rre,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rre.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rre.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rre}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Rre.update_input_task_set({{ur8k_{i+1}_task_instruction_Rre}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Rre.update_input_task_set({{ur8k_{i+1}_task_instruction_Rre}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Rre.update_input_task_set({{ur8k_{i+1}_task_instruction_Rre}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Rre.update_input_task_set({{ur8k_{i+1}_task_instruction_Rre}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BO_A_Rre.update_input_task_set({{ur8k_{i+1}_task_instruction_Rre}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Rre.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Rre}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rre.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rre}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Rre.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Rre}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Rre.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Rre}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rre.update_input_task_set({{ur8k_{i+1}_task_BO_{t}_Rre}}, BaseTaskDependency.FF)")
    # ---


    # ---
    # workflow（水洗い）の作成
    exec(f"ur8k_{i+1}_Rwa = project.create_workflow('Washing Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rwa = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rwa', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rwa}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rwa,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rwa}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rwa.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Rwa = ur8k_{i+1}_Rwa.create_task('Task C2room Urashima #{i+1} {t} Rwa', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rwa}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rwa}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Rwa,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Rwa.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Rwa.update_input_task_set({{ur8k_{i+1}_task_C2_A_Rwa}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rwa = ur8k_{i+1}_Rwa.create_task('Task Hangar Urashima #{i+1} {t} Rwa', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rwa}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rwa}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rwa,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rwa.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rwa.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rwa}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Rwa = ur8k_{i+1}_Rwa.create_task('Task Bridge Urashima #{i+1} {t} Rwa', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rwa}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rwa}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Rwa,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Rwa.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Rwa.update_input_task_set({{ur8k_{i+1}_task_BR_A_Rwa}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rwa = ur8k_{i+1}_Rwa.create_task('Task Deck Urashima #{i+1} {t} Rwa', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rwa}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rwa}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Rwa,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rwa.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rwa.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rwa}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rwa = ur8k_{i+1}_Rwa.create_task('Task Deck Urashima #{i+1} {t} Rwa', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rwa}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rwa}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Rwa,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rwa.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rwa.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rwa}}, BaseTaskDependency.SS)")

    # 作業艇　船員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rwa = ur8k_{i+1}_Rwa.create_task('Task Boat Urashima #{i+1} {t} Rwa', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rwa}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rwa}})")
        exec(f"tasks_boat_t1.append(ur8k_{i+1}_task_BO_{t}_Rwa,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rwa.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rwa.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rwa}}, BaseTaskDependency.SS)")
    for t in "CD":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rwa = ur8k_{i+1}_Rwa.create_task('Task Boat Urashima #{i+1} {t} Rwa', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rwa}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rwa}})")
        exec(f"tasks_boat_t2.append(ur8k_{i+1}_task_BO_{t}_Rwa,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rwa.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rwa.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rwa}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Rwa.update_input_task_set({{ur8k_{i+1}_task_instruction_Rwa}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Rwa.update_input_task_set({{ur8k_{i+1}_task_instruction_Rwa}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Rwa.update_input_task_set({{ur8k_{i+1}_task_instruction_Rwa}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Rwa.update_input_task_set({{ur8k_{i+1}_task_instruction_Rwa}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BO_A_Rwa.update_input_task_set({{ur8k_{i+1}_task_instruction_Rwa}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Rwa.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Rwa}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rwa.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rwa}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Rwa.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Rwa}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Rwa.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Rwa}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rwa.update_input_task_set({{ur8k_{i+1}_task_BO_{t}_Rwa}}, BaseTaskDependency.FF)")
    # ---


    # ---
    # workflow（移動）の作成
    exec(f"ur8k_{i+1}_Rpo = project.create_workflow('Positioning Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rpo = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rpo', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rpo}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rpo,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rpo}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rpo.name: 1.0}})")

    # 司令室　運航員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_C2_{t}_Rpo = ur8k_{i+1}_Rpo.create_task('Task C2room Urashima #{i+1} {t} Rpo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rpo}})")
        exec(f"c2_room.update_targeted_task_set({{ur8k_{i+1}_task_C2_{t}_Rpo}})")
        exec(f"tasks_engineer_t1.append(ur8k_{i+1}_task_C2_{t}_Rpo,)")
        exec(f"ur8k_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_C2_{t}_Rpo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_C2_{t}_Rpo.update_input_task_set({{ur8k_{i+1}_task_C2_A_Rpo}}, BaseTaskDependency.SS)")

    # 格納庫　運航員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rpo = ur8k_{i+1}_Rpo.create_task('Task Hangar Urashima #{i+1} {t} Rpo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rpo}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rpo}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rpo,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rpo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rpo.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rpo}}, BaseTaskDependency.SS)")
    
    # 船橋　船員task
    for t in "A":
        exec(f"ur8k_{i+1}_task_BR_{t}_Rpo = ur8k_{i+1}_Rpo.create_task('Task Bridge Urashima #{i+1} {t} Rpo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rpo}})")
        exec(f"bridge.update_targeted_task_set({{ur8k_{i+1}_task_BR_{t}_Rpo}})")
        exec(f"tasks_work_t1.append(ur8k_{i+1}_task_BR_{t}_Rpo,)")
        exec(f"nav_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BR_{t}_Rpo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BR_{t}_Rpo.update_input_task_set({{ur8k_{i+1}_task_BR_A_Rpo}}, BaseTaskDependency.SS)")

    # 甲板　船員task
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rpo = ur8k_{i+1}_Rpo.create_task('Task Deck Urashima #{i+1} {t} Rpo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rpo}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rpo}})")
        exec(f"tasks_work_t2.append(ur8k_{i+1}_task_DE_{t}_Rpo,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rpo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rpo.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rpo}}, BaseTaskDependency.SS)")
    for t in "EFGH":
        exec(f"ur8k_{i+1}_task_DE_{t}_Rpo = ur8k_{i+1}_Rpo.create_task('Task Deck Urashima #{i+1} {t} Rpo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rpo}})")
        exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_DE_{t}_Rpo}})")
        exec(f"tasks_work_t3.append(ur8k_{i+1}_task_DE_{t}_Rpo,)")
        exec(f"ur8k_utility_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_DE_{t}_Rpo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_DE_{t}_Rpo.update_input_task_set({{ur8k_{i+1}_task_DE_A_Rpo}}, BaseTaskDependency.SS)")

    # 作業艇　船員task
    for t in "AB":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rpo = ur8k_{i+1}_Rpo.create_task('Task Boat Urashima #{i+1} {t} Rpo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rpo}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rpo}})")
        exec(f"tasks_boat_t1.append(ur8k_{i+1}_task_BO_{t}_Rpo,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rpo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rpo.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rpo}}, BaseTaskDependency.SS)")
    for t in "CD":
        exec(f"ur8k_{i+1}_task_BO_{t}_Rpo = ur8k_{i+1}_Rpo.create_task('Task Boat Urashima #{i+1} {t} Rpo', need_facility=True, default_work_amount=5)")

        exec(f"ur8k_{i+1}_recover.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rpo}})")
        exec(f"boat.update_targeted_task_set({{ur8k_{i+1}_task_BO_{t}_Rpo}})")
        exec(f"tasks_boat_t2.append(ur8k_{i+1}_task_BO_{t}_Rpo,)")
        exec(f"boat_console_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_BO_{t}_Rpo.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_BO_{t}_Rpo.update_input_task_set({{ur8k_{i+1}_task_BO_A_Rpo}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_C2_A_Rpo.update_input_task_set({{ur8k_{i+1}_task_instruction_Rpo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_HA_A_Rpo.update_input_task_set({{ur8k_{i+1}_task_instruction_Rpo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BR_A_Rpo.update_input_task_set({{ur8k_{i+1}_task_instruction_Rpo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_DE_A_Rpo.update_input_task_set({{ur8k_{i+1}_task_instruction_Rpo}}, BaseTaskDependency.SS)")
    exec(f"ur8k_{i+1}_task_BO_A_Rpo.update_input_task_set({{ur8k_{i+1}_task_instruction_Rpo}}, BaseTaskDependency.SS)")
    for t in "AB":
        exec(f"ur8k_{i+1}_task_instruction_Rpo.update_input_task_set({{ur8k_{i+1}_task_C2_{t}_Rpo}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rpo.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rpo}}, BaseTaskDependency.FF)")
    for t in "A":
        exec(f"ur8k_{i+1}_task_instruction_Rpo.update_input_task_set({{ur8k_{i+1}_task_BR_{t}_Rpo}}, BaseTaskDependency.FF)")
    for t in "ABCDEFGH":
        exec(f"ur8k_{i+1}_task_instruction_Rpo.update_input_task_set({{ur8k_{i+1}_task_DE_{t}_Rpo}}, BaseTaskDependency.FF)")
    for t in "ABCD":
        exec(f"ur8k_{i+1}_task_instruction_Rpo.update_input_task_set({{ur8k_{i+1}_task_BO_{t}_Rpo}}, BaseTaskDependency.FF)")
    # ---

    
    # 異なるworkflow間のtask依存関係
    exec(f"ur8k_{i+1}_task_instruction_Rsu.update_input_task_set({{ur8k_{i+1}_control_A_Ss7}})")
    exec(f"ur8k_{i+1}_task_instruction_Rca.update_input_task_set({{ur8k_{i+1}_task_instruction_Rsu}})")
    exec(f"ur8k_{i+1}_task_instruction_Rdr.update_input_task_set({{ur8k_{i+1}_task_instruction_Rca}})")
    exec(f"ur8k_{i+1}_task_instruction_Rri.update_input_task_set({{ur8k_{i+1}_task_instruction_Rdr}})")
    exec(f"ur8k_{i+1}_task_instruction_Rho.update_input_task_set({{ur8k_{i+1}_task_instruction_Rri}})")
    exec(f"ur8k_{i+1}_task_instruction_Rli.update_input_task_set({{ur8k_{i+1}_task_instruction_Rho}})")
    exec(f"ur8k_{i+1}_task_instruction_Rsw.update_input_task_set({{ur8k_{i+1}_task_instruction_Rli}})")
    exec(f"ur8k_{i+1}_task_instruction_Rlo.update_input_task_set({{ur8k_{i+1}_task_instruction_Rsw}})")
    exec(f"ur8k_{i+1}_task_instruction_Rla.update_input_task_set({{ur8k_{i+1}_task_instruction_Rlo}})")
    exec(f"ur8k_{i+1}_task_instruction_Rre.update_input_task_set({{ur8k_{i+1}_task_instruction_Rla}})")
    exec(f"ur8k_{i+1}_task_instruction_Rwa.update_input_task_set({{ur8k_{i+1}_task_instruction_Rre}})")
    exec(f"ur8k_{i+1}_task_instruction_Rpo.update_input_task_set({{ur8k_{i+1}_task_instruction_Rwa}})")


    # ---
    # componentの作成
    exec(f"ur8k_{i+1}_maintenance = ur8k.create_component('Urashima #{i+1} Maintenance')")
    # ---


    # ---
    # workflow（整備）の作成
    exec(f"ur8k_{i+1}_Rma = project.create_workflow('Maintaining Urashima #{i+1} R')")
    
    #指示系task
    exec(f"ur8k_{i+1}_task_instruction_Rma = ur8k_{i+1}_instruction.create_task('Task Instruction Urashima #{i+1} Rma', need_facility=True, default_work_amount=0)")
    exec(f"ur8k_{i+1}_instruction_sys.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rma}})")
    exec(f"tasks_instruction.append(ur8k_{i+1}_task_instruction_Rma,)")
    exec(f"deck.update_targeted_task_set({{ur8k_{i+1}_task_instruction_Rma}})")
    exec(f"instruction_tool.workamount_skill_mean_map.update({{ur8k_{i+1}_task_instruction_Rma.name: 1.0}})")

    # 格納庫　運航員task
    for t in "ABCDE":
        exec(f"ur8k_{i+1}_task_HA_{t}_Rma = ur8k_{i+1}_Rma.create_task('Task Hangar Urashima #{i+1} {t} Rma', need_facility=True, default_work_amount=60)")

        exec(f"ur8k_{i+1}_maintenance.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rma}})")
        exec(f"hangar.update_targeted_task_set({{ur8k_{i+1}_task_HA_{t}_Rma}})")
        exec(f"tasks_engineer_t2.append(ur8k_{i+1}_task_HA_{t}_Rma,)")
        exec(f"ur8k_engineering_tool_{t}.workamount_skill_mean_map.update({{ur8k_{i+1}_task_HA_{t}_Rma.name: 1.0}})")
        if t != "A":
            exec(f"ur8k_{i+1}_task_HA_{t}_Rma.update_input_task_set({{ur8k_{i+1}_task_HA_A_Rma}}, BaseTaskDependency.SS)")
    
    #　　taskの依存関係
    exec(f"ur8k_{i+1}_task_HA_A_Rma.update_input_task_set({{ur8k_{i+1}_task_instruction_Rma}}, BaseTaskDependency.SS)")
    for t in "ABCDE":
        exec(f"ur8k_{i+1}_task_instruction_Rma.update_input_task_set({{ur8k_{i+1}_task_HA_{t}_Rma}}, BaseTaskDependency.FF)")
    # ---


    # 異なるworkflow間のtask依存関係
    exec(f"ur8k_{i+1}_task_instruction_Rma.update_input_task_set({{ur8k_{i+1}_task_instruction_Rpo}})")




    if i != 0:
        exec(f"ur8k_{i+1}_task_instruction_Lpr.update_input_task_set({{ur8k_{i}_task_instruction_Rpo}})")
````

````python
# for task in tasks_maneuver_t1+tasks_maneuver_t2:
#     task.worker_priority_rule = ResourcePriorityRuleMode.VC
# for task in tasks_engine_t1+tasks_engine_t2:
#     task.worker_priority_rule = ResourcePriorityRuleMode.VC
# for task in tasks_elec_t1:
#     task.worker_priority_rule = ResourcePriorityRuleMode.VC
for task in tasks_engineer_t1+tasks_engineer_t2:
    task.worker_priority_rule = ResourcePriorityRuleMode.VC
for task in tasks_work_t2+tasks_work_t3:
    task.worker_priority_rule = ResourcePriorityRuleMode.VC
for task in tasks_boat_t1+tasks_boat_t2:
    task.worker_priority_rule = ResourcePriorityRuleMode.VC


import pDESy.model.base_project
from pDESy.model.base_priority_rule import sort_task_list as original_sort_task_list

def create_custom_priority_rule(target_priority_tasks, default_rule_mode=TaskPriorityRuleMode.EST):
    target_ids = set(task.ID for task in target_priority_tasks)
    
    def custom_sorter(task_list):
        high_priority_list = []
        normal_priority_list = []
        for task in task_list:
            if task.ID in target_ids:
                high_priority_list.append(task)
            else:
                normal_priority_list.append(task)
        sorted_normal_list = original_sort_task_list(normal_priority_list, default_rule_mode)
        return high_priority_list + sorted_normal_list
    return custom_sorter

def patched_sort_task_list(task_list, priority_rule_mode=TaskPriorityRuleMode.EST):
    if callable(priority_rule_mode):
        return priority_rule_mode(task_list)
    else:
        return original_sort_task_list(task_list, priority_rule_mode)
    
pDESy.model.base_project.sort_task_list = patched_sort_task_list

priority_tasks = []  
for i in range(number_of_days):
    for t in ['04_1', '48_1', '80_1', '04_2', '48_2', '80_2']:
        exec(f"priority_tasks.append(maneuver_{t}_day{i+1}_A)")
        exec(f"priority_tasks.append(maneuver_{t}_day{i+1}_B)")
        exec(f"priority_tasks.append(eng_{t}_day{i+1}_A)")
        exec(f"priority_tasks.append(eng_{t}_day{i+1}_B)")
        exec(f"priority_tasks.append(elec_{t}_day{i+1}_A)")
for i in range(number_of_ur8k_survey):
    for t in ['Ss1', 'Ss2', 'Ss3', 'Ss4', 'Ss5', 'Ss6', 'Ss7']:
        exec(f"priority_tasks.append(ur8k_{i+1}_control_A_{t})")
        exec(f"priority_tasks.append(ur8k_{i+1}_control_B_{t})")
for i in range(number_of_sn6500_survey):
    for t in ['Ss1', 'Ss2', 'Ss3', 'Ss4',]:
        exec(f"priority_tasks.append(sn6500_{i+1}_monitor_A_{t})")
        exec(f"priority_tasks.append(sn6500_{i+1}_monitor_B_{t})")
        exec(f"priority_tasks.append(sn6500_{i+1}_monitor_C_{t})")
        exec(f"priority_tasks.append(sn6500_{i+1}_control_A_{t})")
        exec(f"priority_tasks.append(sn6500_{i+1}_control_B_{t})")

custom_rule = create_custom_priority_rule(
    target_priority_tasks=priority_tasks,
    default_rule_mode=TaskPriorityRuleMode.EST
)

project.simulate(max_time=5761, progress_bar=True, task_priority_rule=custom_rule)
````
